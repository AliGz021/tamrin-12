```python
import os
import time
```

```python
folder_name = input('Enter folder name: ')

if os.path.exists(folder_name):
    os.rmdir(folder_name)
else:
    os.mkdir(folder_name)
```

```python
path = input('Enter path: ')

folders_names = input('Enter folders names: ').split()

for folder in folders_names:
    full_path = os.path.join(path, folder)

    if not os.path.exists(full_path):
        os.mkdir(full_path)
```

```python
file_path = input('Enter file path: ')

file_size = os.path.getsize(file_path)

print(file_size)
```

```python
watch_path = input('Enter path to monitor: ')

initial_files = os.listdir(watch_path)

while True:
    time.sleep(5)

    current_files = os.listdir(watch_path)

    new_files = [file for file in current_files if file not in initial_files]
    deleted_files = [file for file in initial_files if file not in current_files]

    if new_files:
        print('New file added:', new_files)

    if deleted_files:
        print('File deleted:', deleted_files)

    initial_files = current_files
```

```python
sort_path = input('Enter folder path: ')

files = os.listdir(sort_path)

folders = ['image', 'audio', 'video', 'document']

for folder in folders:
    folder_path = os.path.join(sort_path, folder)

    if not os.path.exists(folder_path):
        os.mkdir(folder_path)

for file in files:
    if file.endswith(('.jpg', '.jpeg', '.png', '.gif')):
        destination = 'image'

    elif file.endswith(('.mp3', '.wav', '.flac')):
        destination = 'audio'

    elif file.endswith(('.mp4', '.avi', '.mkv')):
        destination = 'video'

    elif file.endswith(('.pdf', '.docx', '.txt', '.xlsx', '.pptx')):
        destination = 'document'

    else:
        continue

    source = os.path.join(sort_path, file)
    destination_path = os.path.join(sort_path, destination)
    destination_file = os.path.join(destination_path, file)

    os.rename(source, destination_file)
```
